# 00 · Setup monitoring schema

Foundation step. Creates the derived schema that every later stage writes to.
**Source / gold stays read-only** — all derived state lives here.

Reusable across businesses: only the **Config** cell changes.

## Config
Set your target catalog/schema. Defaults point at the CoffCoff reference implementation.

```python
dbutils.widgets.text("target_catalog", "mickmickstableaws_catalog", "Target catalog")
dbutils.widgets.text("monitoring_schema", "monitoring", "Derived/monitoring schema")

TARGET_CATALOG = dbutils.widgets.get("target_catalog").strip()
MONITORING_SCHEMA = dbutils.widgets.get("monitoring_schema").strip()
FQ_SCHEMA = f"{TARGET_CATALOG}.{MONITORING_SCHEMA}"

print(f"Derived state target: {FQ_SCHEMA}")
```

## Create the derived schema
Gold/source is never modified — we only create a separate schema for derived state.

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {FQ_SCHEMA} COMMENT 'Derived agentic supply chain state: twins, KPIs, events. Source/gold remains read-only.'")
print(f"Schema ready: {FQ_SCHEMA}")
```

## Verify
Confirm the schema exists (it will have no tables yet).

```python
display(spark.sql(f"SHOW TABLES IN {FQ_SCHEMA}"))
```
